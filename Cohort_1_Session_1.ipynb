**Qiskit Demo 1: Creating Your First Quantum Circuit**

In this demo, we'll create our very first quantum circuit using Qiskit. We'll initialize a single qubit, apply a Hadamard (H) gate to place it into superposition, measure the qubit, and visualize the circuit. This introduces the basic workflow of designing a quantum circuit in Qiskit.

```python
# Import Qiskit
from qiskit import QuantumCircuit

# Create a quantum circuit with 1 qubit and 1 classical bit
qc = QuantumCircuit(1, 1)

# Apply a Hadamard Gate
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Draw the circuit
qc.draw("mpl")
```

**Qiskit Demo 2: Running the Circuit on a Simulator**

Designing a circuit is only the first step. In this demo, we'll execute the circuit using Qiskit's Aer Simulator. By running the circuit multiple times (called shots), we'll observe the measurement outcomes and see how a qubit in superposition produces approximately equal probabilities of measuring 0 and 1.

```python
# Install Qiskit (Run this only once in Google Colab)
#!pip -q install qiskit qiskit-aer matplotlib

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a circuit
qc = QuantumCircuit(1, 1)

# Apply Hadamard Gate
qc.h(0)

# Measure the qubit
qc.measure(0, 0)

# Create simulator
simulator = AerSimulator()

# Execute the circuit 1000 times
job = simulator.run(qc, shots=1000)

# Get the results
result = job.result()

# Count measurement outcomes
counts = result.get_counts()

print("Measurement Results:")
print(counts)

# Draw the circuit
qc.draw("mpl")
```

**Qiskit Demo 3: Understanding Quantum Gates**

Quantum gates manipulate the state of qubits. In this demo, we'll compare three simple circuits: one with no gate, one with the Pauli-X (NOT) gate, and one with the Hadamard gate. This helps us understand how different quantum gates affect the state and measurement of a qubit.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# -------------------------
# Circuit 1 : No Gate
# -------------------------
qc1 = QuantumCircuit(1,1)

qc1.measure(0,0)

result1 = simulator.run(qc1, shots=1000).result()
counts1 = result1.get_counts()

print("No Gate:")
print(counts1)

# -------------------------
# Circuit 2 : X Gate
# -------------------------
qc2 = QuantumCircuit(1,1)

qc2.x(0)

qc2.measure(0,0)

result2 = simulator.run(qc2, shots=1000).result()
counts2 = result2.get_counts()

print("\nX Gate:")
print(counts2)

# -------------------------
# Circuit 3 : Hadamard Gate
# -------------------------
qc3 = QuantumCircuit(1,1)

qc3.h(0)

qc3.measure(0,0)

result3 = simulator.run(qc3, shots=1000).result()
counts3 = result3.get_counts()

print("\nHadamard Gate:")
print(counts3)
```

**Qiskit Demo 4: Creating a Bell State (Quantum Entanglement)**

Quantum entanglement is one of the most fascinating features of quantum computing. In this demo, we'll create a Bell State by combining a Hadamard gate with a CNOT gate. We'll observe how two qubits become entangled and produce correlated measurement outcomes.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a simulator
simulator = AerSimulator()

# Create a circuit with 2 qubits and 2 classical bits
qc = QuantumCircuit(2, 2)

# Step 1: Put the first qubit into superposition
qc.h(0)

# Step 2: Entangle both qubits
qc.cx(0, 1)

# Step 3: Measure both qubits
qc.measure([0, 1], [0, 1])

# Draw the circuit
display(qc.draw("mpl"))

# Run the circuit
job = simulator.run(qc, shots=1000)

result = job.result()

counts = result.get_counts()

print(counts)
```

**PennyLane Demo 1: Your First PennyLane Circuit**

In this demo, we'll build our first quantum circuit using PennyLane. We'll create a quantum device, define a QNode, apply a Hadamard gate, and return the measurement probabilities. This demonstrates PennyLane's Pythonic approach to quantum programming.

```python
import pennylane as qml

# Create a quantum device with 1 qubit
dev = qml.device("default.qubit", wires=1)

# Define a quantum function
@qml.qnode(dev)
def circuit():
    # Apply Hadamard Gate
    qml.Hadamard(wires=0)

    # Return measurement probabilities
    return qml.probs(wires=0)

# Execute the circuit
print(circuit())
```

**PennyLane Demo 2: Understanding Quantum Gates**

Just like in Qiskit, quantum gates determine how qubits evolve. In this demo, we'll compare the effects of no gate, the Pauli-X gate, and the Hadamard gate. By examining the output probabilities, we'll understand how each gate changes the quantum state.

```python
import pennylane as qml

# Create a quantum device
dev = qml.device("default.qubit", wires=1)

# ----------------------------------
# Circuit 1 : No Gate
# ----------------------------------
@qml.qnode(dev)
def no_gate():
    return qml.probs(wires=0)

# ----------------------------------
# Circuit 2 : X Gate
# ----------------------------------
@qml.qnode(dev)
def x_gate():
    qml.PauliX(wires=0)
    return qml.probs(wires=0)

# ----------------------------------
# Circuit 3 : Hadamard Gate
# ----------------------------------
@qml.qnode(dev)
def hadamard_gate():
    qml.Hadamard(wires=0)
    return qml.probs(wires=0)

# Display Results
print("No Gate:")
print(no_gate())

print("\nX Gate:")
print(x_gate())

print("\nHadamard Gate:")
print(hadamard_gate())
```

**PennyLane Demo 3: Creating a Bell State (Quantum Entanglement)**

In this demo, we'll create a Bell State using PennyLane to demonstrate quantum entanglement. By applying a Hadamard gate followed by a CNOT gate, we'll observe how two qubits become correlated. This example mirrors the Qiskit implementation, showing that while the syntax differs, the underlying quantum concepts remain the same.

```python
import pennylane as qml

# Create a device with 2 qubits
dev = qml.device("default.qubit", wires=2)

# Create a quantum circuit
@qml.qnode(dev)
def bell_state():

    # Step 1: Create superposition
    qml.Hadamard(wires=0)

    # Step 2: Entangle both qubits
    qml.CNOT(wires=[0, 1])

    # Return measurement probabilities
    return qml.probs(wires=[0,1])

# Execute the circuit
print(bell_state())
```

**PennyLane Demo 4 – Variational Quantum Circuit (Parameterized Quantum Circuit)**

Unlike the previous circuits, which always behaved the same way, this circuit contains a trainable parameter. By changing the angle of a quantum rotation gate, we can control the probability of measuring different quantum states. This is the foundation of Variational Quantum Algorithms (VQAs) and Quantum Machine Learning, where classical optimization algorithms adjust these parameters to improve the model's performance.

```python
import pennylane as qml
import numpy as np

# Create a quantum device
dev = qml.device("default.qubit", wires=1)

# Define a parameterized quantum circuit
@qml.qnode(dev)
def variational_circuit(theta):

    # Rotation around the Y-axis
    qml.RY(theta, wires=0)

    # Return probabilities
    return qml.probs(wires=0)

# Test different rotation angles
angles = [0, np.pi/4, np.pi/2, np.pi]

for angle in angles:
    print(f"θ = {angle:.2f} radians")
    print(variational_circuit(angle))
    print("-" * 30)
```
